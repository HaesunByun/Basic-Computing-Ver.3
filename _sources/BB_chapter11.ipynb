    - index_col: 지정한 열(컬럼)을 인덱스로 사용한다.
    - usecols: 읽어올 열(컬럼)을 지정한다.

    - 이 외에도 다양한 옵션을 제공한다.

- excel파일을 불러올 때
  - pd.read_excel('파일명')


:::{admonition} 인코딩 & 디코딩
:class: tip  
- 인코딩(encoding)는 코드화 또는 암호화로 불리우며, 문자열을 바이트코드로 변환하는 것을 의미한다.
- 인코딩을 하는 이유는 언어마다 표준화된 규격으로 문자집합을 만들고, 이러한 문자집합을 가지고 정보의 형태 표준화, 보안, 저장 공간 절약 등의 다양한 목적으로 부호화하여 사용하기 위한 것이다.
- 파이썬에서는 문자열을 유니코드로 처리하는데 유니코드를 utf-8, euc-kr, ascii 형식의 인코딩방식을 사용하여 바이트코드로 변환한다. 예를 들어,
```
x='한글'
encoded = x.encode('utf-8')
print(f'Output: {encoded}')
```
Output: b'\xed\x95\x9c\xea\xb8\x80'

- 여기서 결과로 나온 b'\xed\x95\x9c\xea\xb8\x80'가 바이트 코드이다.

- 디코딩(decoding)은 역코드화 또는 복호화로 불리우며, 바이트코드를 문자열로 변환하는 것을 의미한다.
- 이때 디코딩에서도 인코딩 시 사용했던 것과 동일한 인코딩 방식(utf-8, euc-kr, ascii 등)이 사용되어야 한다.
```
decoded = encoded.decode('utf-8')
print(f'Output: {decoded}')
```
Output: 한글

## 자료형 다루기
- .astype(자료형)
  - 대표적인 자료형으로는 int(정수), float(실수), str(문자열), bool(논리값)이 있음
  - astype()은 변환된 새로운 객체를 반환하므로, 결과를 저장하려면 다시 변수에 할당해야 함

````python
import pandas as pd
scores = [[84,87,78], [21,15,84], [87,84,76], [100,87,99],[59,99,59],[46,77,56]]
names=['철수','영이','길동','미영','순이','철이']
lectures=['국어','수학','영어']
d1 = pd.DataFrame(scores, index=names, columns=lectures)
d1
````

````python
d2 = d1.astype('str')
d2
````

````python
d2.sum()
````

## 그룹핑하기

- groupby()는 하나 이상의 열을 기준으로 데이터를 그룹화한 후, 그룹별 통계값을 계산할 때 사용한다.

```
데이터프레임명.groupby('기준열') # 단일 기준으로 그룹
데이터프레임명.groupby(['컬럼1', '컬럼2'])  # 다중 기준으로 그룹
```

- groupby() 자체는 결과를 보여주지 않고, 함수를 이어 적용해야 결과 생성
  - df.groupby('반').size()          # 반별 행의 개수, 결측값과 관계없이 행의 개수를 계산
  - df.groupby('반')['점수'].count() # 반별 점수 데이터의 개수, 점수 열의 결측값을 제외하고 개수를 계산
  - df.groupby('반')['점수'].sum()   # 반별 점수의 합
  - df.groupby('반')['점수'].mean()  # 반별 점수의 평균

````python
d1['반']=['01', '01', '02', '02', '03', '03']
d1['성별']=['남','여','남','여','여','남']
d1
````

````python
d1.groupby('반').sum()
````

````python
d1.groupby('반').mean() #에러
````

````python
d1.groupby('반')['국어'].mean()
````

````python
d1.groupby(['반','성별'])['국어'].mean()
````

## 누락값 처리하기


### 누락 데이터 확인
  - isnull(): 누락 데이터이면 True, 유효한 데이터이면 False를 반환
  - notnull():유효한 데이터이면 True, 누락데이터이면 False를 반환

````python
import pandas as pd
scores = [[84,87,78], [21,15,84], [87,84,76], [100,87,99],[59,99,59],[46,77,56]]
names=['철수','영이','길동','미영','순이','철이']
lectures=['국어','수학','영어']
d1 = pd.DataFrame(scores, index=names, columns=lectures)
d1
````

````python
scores = [[84,87,78], [21,15,84], [87,84,76], [100,87,99],[59,99,59],[46,77,56]]
names=['철수','영이','길동','슬기','순이','철이']
lectures=['국어','수학','영어']
d2 = pd.DataFrame(scores, index=names, columns=lectures)
d2
````

- d1과 d2를 더해서 일부러 누락값을 만들어서 테스트해보자.

````python
d3=d1+d2
d3
````

````python
d3.isnull()
#d3.notnull()
````

- 누락데이터의 개수를 확인하려면 .isnull().sum()으로 null인 데이터의 개수를 합한다.
- 결측치가 하나라도 있는 행을 조회하려면 .any()를 사용한다.

````python
d3.isnull().sum()
````

````python
d3[d3.isnull().any(axis=1)]
````

### 누락 데이터 제거
  - 행제거: .dropna(subset=['열1','열2'], axis=0)
    - '열1' 또는 '열2' 중 하나라도 NaN이면 해당 행(row)을 제거
    - subset 옵션을 생략하면 모든 열을 기준으로 NaN이 있는 행을 제거
    - 디폴트는 axis=0이므로 생략해도 됨
  - 열제거: .dropna(axis=1)
    - NaN이 하나라도 있는 열 전체를 제거
    - .dropna(axis=1)은 거의 사용하지 않음

````python
d4=d3.dropna(axis=0)
d4
````

### 누락 데이터 치환
  - .fillna(값)

````python
min_kor = d3['국어'].min()
d3['국어'] = d3['국어'].fillna(min_kor)
d3
````

## 문자열 데이터 처리

데이터 분석에서는 아래와 같이 문자열 데이터를 분석에 맞게 다듬는 다양한 작업을 수행

 - 포함 여부 확인

 - 문자열 치환/삭제

 - 대소문자 변환

 - 문자열 분리/합치기

 문자열 처리를 통해 정제된 데이터를 기반으로 분석 및 통계 작업을 수행할 수 있음

문자열 데이터는 <b>Series.str 접근자(accessor)</b>를 사용해 처리

  - 파이썬의 문자열 메서드를 벡터화하여 열 전체를 한 번에 처리  
     
- 문자열 관련 메서드는 모두 `str.메서드()` 형태로 사용

|                  | 메서드                           | 설명                  |
| ------------------ | ----------------------------- | ------------------- |
| **대·소문자 변환**       | `str.upper()`                 | 모든 문자를 대문자로         |
|                    | `str.lower()`                 | 모든 문자를 소문자로         |
|                    | `str.title()`                 | 각 단어 첫 글자 대문자로      |
| **공백/문자 제거**       | `str.strip()`                 | 앞뒤 공백 제거            |
|                    | `str.lstrip()`                | 왼쪽 공백 제거            |
|                    | `str.rstrip()`                | 오른쪽 공백 제거           |
| **검색(조건)**         | `str.contains('a')`           | 특정 문자열 포함 여부        |
|                    | `str.startswith('A')`         | 특정 문자로 시작           |
|                    | `str.endswith('.com')`        | 특정 문자로 끝남           |
| **추출 / 인덱싱**       | `str.get(i)`                  | i번째 문자 가져오기         |
|                    | `str.slice(start, end)`       | 부분 문자열 추출           |
|                    | `str[0:3]`                    | 파이썬 슬라이싱 지원         |
| **치환**             | `str.replace('a','b')`        | 문자열 치환 (regex 가능)   |
| **분리(split) / 확장** | `str.split('-')`              | 구분자 기준 분리 (리스트)     |
|                    | `str.split('-', expand=True)` | 여러 열로 확장            |
| **결합(cat)**        | `str.cat(sep='-')`            | 문자열 결합              |
| **길이**             | `str.len()`                   | 문자열의 길이 반환          |
| **정규표현식 처리**       | `str.findall(r'\d+')`         | 패턴과 일치하는 모든 값 리스트   |
|                    | `str.extract(r'(\d+)')`       | 첫 번째 일치값 추출 (캡처 그룹) |


````python
# 예제 데이터
data = {
    'Name': ['Alice', '  Bob', 'Charlie', '  David  '],
    'Business Name': ['Blue Coffee', 'Morning Coffee', 'Fresh Bakery', 'Green Mart']
}

df = pd.DataFrame(data)
print(df)
````

````python
# Business Name에 'coffee' 포함 여부 확인
df['coffee_brand'] = df['Business Name'].str.contains('coffee')
print(df)
````
