### 선그래프: plt.plot()
- **선 그래프**는 수량을 점으로 표시하고 점과 점 사이의 거리를 직선으로 연결한 그래프이다.
- 연속하는 데이터 값들을 직선 또는 곡선으로 연결하여 데이터 값 사이의 관계를 나타낸다.
- 시간에 따른 데이터의 변화 추세를 파악하는데 유용하다.
```
plt.plot(x, y, 꾸미기 옵션)
```


- 선그래프에 사용될 수 있는 다양한 꾸미기 옵션이 있다.
| 구분 | 옵션                        | 설명        | 예시                                |
| -- | ------------------------- | --------- | --------------------------------- |
| 선  | `color`                   | 선 색상      | `'r'`, `'blue'`, `'#FF5733'`      |
| 선  | `linestyle` (`ls`)        | 선 스타일     | `'-'`, `'--'`, `':'`, `'-.'`      |
| 선  | `linewidth` (`lw`)        | 선 두께      | `1`, `2.5`, `4`                   |
| 선  | `alpha`                   | 투명도 (0~1) | `0.3`, `0.7`                      |
| 선  | `zorder`                  | 겹침 순서     | `1`, `2`                          |
| 마커 | `marker`                  | 마커 모양     | `'o'`, `'s'`, `'^'`, `'x'`, `'*'` |
| 마커 | `markersize` (`ms`)       | 마커 크기     | `5`, `10`                         |
| 마커 | `markerfacecolor` (`mfc`) | 마커 내부 색   | `'red'`, `'white'`                |
| 마커 | `markeredgecolor` (`mec`) | 마커 테두리 색  | `'black'`                         |
| 마커 | `markeredgewidth` (`mew`) | 테두리 두께    | `1`, `2`                          |
| 기타 | `label`                   | 범례 이름     | `'실험 데이터'`                        |
| 기타 | `solid_capstyle`          | 실선 끝 모양   | `'round'`, `'butt'`               |
| 기타 | `dash_capstyle`           | 점선 끝 모양   | `'round'`                         |

#### 다중 선그래프
- 여러 개의 그래프를 하나의 도화지에 그리고 싶을 때에는 그래프를 먼저 그리고 가장 마지막에 plt.show()하면 된다.

````python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 생성
x = np.arange(1, 11)
y1 = x * 2
y2 = x ** 1.5
y3 = x + 5

# 다중 선 그래프
# 모든 선에는 의미를 구분할 수 있도록 각각 'Linear', 'Power', 'Shifted'라는 label을 지정
# 첫 번째 선은 실선(linestyle='-')과 원형 마커(marker='o')
# 두 번째 선은 점선(linestyle='--')과 사각형 마커(marker='s')
# 세 번째 선은 점선(linestyle=':')과 삼각형 마커(marker='^')
# Todo


# 그래프 꾸미기
plt.title('Multiple Line Plot Example') # 제목
plt.xlabel('X value') # x축 라벨
plt.ylabel('Y value') # y축 라벨
plt.legend() # 범례
plt.grid(True)

plt.show()
````

### 산포도(산점도) 그래프: plt.scatter()
- **산포도그래프**는 도표 위에 두 변수 x와 y 값이 만나는 지점을 표시한다.
- 이 그래프를 통해 두 변수 사이의 관계를 알 수 있다.

```

```

- 산포도 그래프에 사용되는 옵션은 다음과 같다.
| 구분  | 옵션             | 설명           | 예시                           |
| --- | -------------- | ------------ | ---------------------------- |
| 기본  | `s`            | 점 크기 (면적 기준) | `20`, `100`                  |
| 기본  | `c`            | 점 색상         | `'r'`, `'blue'`, `'#1f77b4'` |
| 기본  | `marker`       | 점 모양         | `'o'`, `'s'`, `'^'`, `'x'`   |
| 기본  | `alpha`        | 투명도 (0~1)    | `0.3`, `0.7`                 |
| 기본  | `label`        | 범례 이름        | `'Class A'`                  |
| 테두리 | `edgecolors`   | 점 테두리 색      | `'black'`, `'none'`          |
| 테두리 | `linewidths`   | 점 테두리 두께     | `0.5`, `1.5`                 |
| 색상  | `cmap`         | 컬러맵 (연속값일 때) | `'viridis'`, `'plasma'`      |

https://matplotlib.org/tutorials/colors/colormaps.html

## Matplotlib: 판다스 데이터 활용

````python
import pandas as pd
scores = [[84, 87, 78], [21, 15, 84], [87, 84, 76], [100, 87, 99], [59, 99, 59], [46, 77, 56]]
names = ['철수', '영이', '길동', '미영', '순이', '철이']
lectures = ['국어', '영어', '수학']
df=pd.DataFrame(scores, index=names, columns=lectures)
df
````

- 데이터프레임으로 그래프를 그리는 방법은 다음과 같이 두 가지를 사용할 수 있다.
  - 첫 번째는 plt.그래프함수명(x, y)와 같이 앞서 설명했던 방법으로 그리는 것이다.
    - 예) plt.plot(df.index, df['국어'])
  - 두 번째는 데이터프레임명.plot(kind='그래프종류')와 같이 데이터프레임명을 .앞에 사용하는 것이다.
    - 예) df['국어'].plot(kind='line')
    - kind 옵션은 다음과 같다.
      - 'line': 선그래프
      - 'bar': 수직막대그래프
      - 'barh': 수평막대그래프
      - 'pie': 원그래프
      - 'box': 박스그래프
      - 'kde': 커널밀도그래프

### 여러개 그래프: plt.subplot()
- 여러 개의 그래프를 하나의 도화지에 나타내보자.

### 파이 그래프: df변수명.plot(kind='pie')
- autopct : 파이조각을 백분율로 나타낸다.
- explode: 특정 조각을 돌출시킨다.

```
explode = [0.0, 0.1, 0.0, 0.0]
데이터프레임명.plot(kind='pie',
                 figsize=(7,5),
                 autopct='%.1f%%',
                 explode = explode,
                 colors=['chocolate','bisque','cadetblue','yellow'])
```

### 카테고리별 회귀선이 있는 산포도 그리기: sns.lmplot()
- regplot()과 비슷하나, hue 파라미터를 지정하여 카테고리별로 각각 데이터의 분포와 선형 관계를 표시할 수 있다.


😄 학습 방식으로 구분하여 공부 시간과 시험 점수와의 관계를 보기위해 회귀선이 있는 산포도 그래프를 그려보자.

### 화면분할 그래프: .FacetGrid()
- **FacetGrid그래프**는 행, 열 방향으로 서로 다른 조건을 적용하여 여러 개의 서브 플롯을 만든다.
- map() 함수를 이용하여 각 서브 플롯에 적용할 그래프 종류를 전달한다.


😄 학습 방식을 열로, Pass/Fail을 행으로 나누어 각 서브 플롯에 공부 시간과 시험 점수에 대한 회귀선을 가진 산포도 그래프를 그려보자.

````python
score_df['Pass/Fail'] = ['Pass' if i >= 75 else 'Fail' for i in score_df['시험 점수']]
score_df.head()
````

````python
g=sns.FacetGrid (data=score_df, col='학습 방식', row='Pass/Fail')
g=g.map(sns.regplot, '공부 시간', '시험 점수')
````
