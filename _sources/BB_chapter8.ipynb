# 중간고사



## 시험 일정 및 범위

📌시험 일시: 별도 공지

📌시험 장소 및 방법: 26동 102호, 대면 시험 원칙

📌시험 범위: Chapter 1 ~ Chapter 7

📌시험 유형:객관식, 단답형 코딩, 프로그래밍 문제

📌 프로그래밍 문제 난이도

- 이론 수업에서 다루는 문제를 그대로 내거나 변형
- 실습 수업의 사전/도전/완전 실습 문제를 그대로 내거나 변형
- 시험 범위에 해당하는 개념을 다루는 새로운 문제

## 시험 절차

- 출석 체크 및 시험 준비
- 준비 완료 후 엘리스(Ellice) 시험 문제 공개
- 시험 진행
- 시험 종료 (엘리스 자동 마감)


## 시험 유의사항 안내

- 개인 노트북 사용 가능합니다.
- 엘리스(Ellice)에서만 진행 가능합니다.(Google Colab 사용 금지)
- 웹 이론 강의자료 사이트 보기 허용합니다. 단, 그 밖의 인터넷 검색은 할 수 없습니다.
- 신분증 검사가 있습니다. 마스크를 착용한 경우, 얼굴 확인 시 잠시 마스크를 내려주시기 바랍니다.
- 시험 시작 후 지각 입장 시 상황에 따라 점수 감점이 있을 수 있습니다. 지각하지 않도록 주의해주세요.
- 코딩 툴은 엘리스(Ellice)만 사용 가능하며, AI 기능은 사용할 수 없습니다.
- 시험 당일 고정 좌석이 지정됩니다. 데스크톱을 사용하는 경우, 미리 확인하시기 바랍니다.
- 노트북 등의 시험 기기는 미리 충전하거나 충전기를 지참해 주세요.
- 시험 중 컴퓨터 문제 또는 네트워크 연결 끊김 발생 시 당황하지 말고 조용히 손을 들어 알려주시기 바랍니다.
- Wi-Fi 연결 불가, 엘리스 접속 불가 등의 문제가 발생할 경우 시험이 일시 중단됩니다. 개인 핫스팟을 사용해 인터넷 연결이 가능한 경우라도 모든 응시자는 키보드에서 손을 내려놓고 대기해야 합니다.
- 다른 프로그램 또는 파일을 열거나, 웹 이론 강의 자료 외 인터넷 검색, 메신저 프로그램 등을 사용할 경우 부정행위로 간주됩니다.
- 엘리스에서 Key Stroking 기능(키 입력 로그 기록)이 활성화되고 모든 응시자의 화면이 녹화됩니다.
- 엘리스에 저장된 최종 코드 기준으로 채점되며, 일부 문제는 부분 채점이 적용될 수 있습니다.
- "테스트 종료하기" 버튼을 누르면 재응시가 불가합니다. 응시를 확실히 마친 경우에만 종료 버튼을 눌러주세요. 버튼을 누르지 않아도 응시 기간이 끝나면 자동으로 종료됩니다.
- 시험 종료 전 퇴실을 원할 경우, '테스트 종료하기' 버튼을 누른 것을 감독관이 확인해야 퇴실 할 수 있습니다.
- 질문이 있을 경우 조용히 손을 들어주세요. 다만, 공정성을 해칠 가능성이 있는 질문에는 답변하지 않을 수 있습니다.
- 필요 시, 빈 A4 용지 사용이 가능합니다.

🚨모든 유의 사항은 시험 운영의 공정성을 확보하기 위한 조치이므로 반드시 준수해 주시기 바랍니다. 시험 규정을 위반할 경우 불이익이 있을 수 있으니, 안내 사항을 숙지하고 성실히 응시해 주시기 바랍니다.🚨
