#문제1. 좌석 예약 현황 출력

: 영화관 전체 좌석 목록과 이미 예약된 좌석 목록을 함수 밖에서 입력받아 저장한 뒤, 좌석 예약 상태를 관리하는 seat_check(seats, reserved) 함수를 정의하고 호출하시오.


***seat_check(seats, reserved)* 함수**
1. 반드시 함수로 정의해야 한다.
2. seats와 reserved를 매개변수로 받아야 한다.
3. 전체 좌석을 반복문으로 확인해야 한다.
4. 예약된 좌석이면 "예약됨", 아니면 "예약 가능"을 출력해야 한다.
5. 마지막에 예약 가능한 좌석만 모은 리스트를 출력해야 한다.
6. 예약 가능한 좌석 개수도 함께 출력해야 한다.

<font color='blue'>`test case-1`</font>

Input Sample

```
A1 A2 A3 B1 B2 B3
A2 B1
```


Output Sample
```
영화관 좌석:
예약된 좌석:
A1 : 예약 가능
A2 : 예약됨
A3 : 예약 가능
B1 : 예약됨
B2 : 예약 가능
B3 : 예약 가능
예약 가능 좌석: ['A1', 'A3', 'B2', 'B3']
예약 가능 좌석 수: 4
```


<font color='blue'>`test case-2`</font>

Input Sample

```
C1 C2 C3 D1
C1 D1
```


Output Sample
```
영화관 좌석:
예약된 좌석:
C1 : 예약됨
C2 : 예약 가능
C3 : 예약 가능
D1 : 예약됨
예약 가능 좌석: ['C2', 'C3']
예약 가능 좌석 수: 2
```

<font color='blue'>`test case-3`</font>

Input Sample

```
E1 E2 E3
E2
```


Output Sample
```
영화관 좌석:
예약된 좌석:
E1 : 예약 가능
E2 : 예약됨
E3 : 예약 가능
예약 가능 좌석: ['E1', 'E3']
예약 가능 좌석 수: 2
```

````python
def seat_check(seats, reserved):
  available = []
  for seat in seats:
    if seat in reserved:
      print(seat, ':예약됨')
    else:
      print(seat,':예약 가능')
      available.append(seat)

  print('예약 가능 좌석:', available)
  print('예약 가능 좌석 수:', len(available))

seats = input('영화관 좌석: ').split()
reserved = input('예약된 좌석: ').split()

seat_check(seats, reserved)
````

단, 좌석 현황 출력은 문제 2에서 정의한 seat_check(seats, reserved) 함수를 활용해야 한다.

````python
def seat_check(seats, reserved):
  available = []
  for seat in seats:
    if seat in reserved:
      print(seat, ':예약됨')
    else:
      print(seat,':예약 가능')
      available.append(seat)

  print('예약 가능 좌석:', available)
  print('예약 가능 좌석 수:', len(available))

def reserved_seat(seats, reserved):
  seat = input('원하는 좌석을 입력하세요: ')

  if seat in reserved:
    print('이미 예약된 좌석입니다.')
  else:
    reserved.append(seat)
    print('예약이 완료되었습니다.')

  seat_check(seats, reserved)

seats = input('영화관 좌석: ').split()
reserved = input('예약된 좌석: ').split()

reserved_seat(seats, reserved)
````

# 문제3. 택배 박스 최소 개수 구하기

아래의 기능을 가지고 있는 item_input() 함수와 count_boxes() 함수를 정의하고 호출하는 코드를 작성하시오.

***item_input()*** 함수

1. 사용자로부터 포장할 물건의 개수를 입력받아야 한다.
2. 입력된 값이 1 이상 200 이하의 정수가 아니면 재입력을 요구한다.
3. 올바르게 입력된 물건 개수는 반환한다.

***count_boxes()*** 함수
1. 박스의 개수를 계산할 물건 개수를 전달받아야 한다.
2. 사용할 수 있는 박스의 크기는 20개용, 10개용, 5개용, 1개용이다.
3. 각 박스는 무한히 있다고 가정한다.
4. 반복문을 사용하여 최소로 필요한 박스의 개수를 계산해야 한다.
5. 각 박스별 개수와 총 박스 개수를 출력해야 한다.

Input Sample

Output Sample

```
포장할 물건 개수:
필요한 박스 개수
20개용 박스: 2개
10개용 박스: 0개
5개용 박스: 1개
1개용 박스: 2개
총 박스 개수: 5개
```

```
잘못된 입력입니다. 1 이상 200 이하의 정수를 입력하세요.
포장할 물건 개수:
필요한 박스 개수

```



Output Sample
```
포장할 물건 개수:
필요한 박스 개수
20개용 박스: 5개
10개용 박스: 0개
5개용 박스: 0개
1개용 박스: 0개
총 박스 개수: 5개
```

````python
def item_input():
  while True:
    count =  int(input('포장할 물건 개수: '))
    if 1<=count and count<=200:
      return count
    else:
      print('잘못된 입력입니다. 1 이상 200 이하의 정수를 입력하세요.')

def count_boxes(count):
  box_size=[20,10,5,1]
  total_boxes=0
  print('필요한 박스 개수')
  for size in box_size:
    num = count // size
    print(f'{size}개용 박스: {num}개')
    count %= size
    total_boxes += num
  print(f'총 박스 개수: {total_boxes}개')

count = item_input()
count_boxes(count)
````
